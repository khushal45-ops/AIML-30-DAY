```python
def average_marks(students):
  if not students:
    return None

  marks_list = [s["marks"] for s in students if "marks" in s]

  if not marks_list:
    return None

  return sum(marks_list) / len(marks_list)

print(average_marks([{"name": "A", "marks": 80}, {"name": "B", "marks": 90}]))
print(average_marks([]))
print(average_marks([{"name": "A", "marks": 80}, {"name": "B"}]))
print(average_marks([{"name": "A"}, {"name": "B"}]))
```

An empty list has no average, and a student with no marks also gives no average. So the function returns None in both cases instead of 0, which could be mistaken for a real score.

```python
import seaborn as sns
df = sns.load_dataset("titanic")
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.groupby("pclass", observed=True)["deck"].apply(lambda x: x.isna().mean())
```

```python
df["age"] = df.groupby("pclass")["age"].transform(lambda x: x.fillna(x.median()))
df["age"].isna().sum()
```

```python
df.groupby("pclass")["fare"].mean()
```

The average fare for 1st class is higher because upper decks had larger and better cabins, and most wealthy passengers stayed there.

# **Summary**
What I did: I filled missing age values using the median age of each passenger class, and checked the missing percentage of the deck column.
Why: A global median would ignore the difference between classes, and deck was 77% missing and strongly linked to class, so I decided to drop it for modeling.
What I learned: Before filling or dropping missing data, I need to check whether the missing values follow a pattern, not just count how many are missing.
